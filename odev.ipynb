```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("data.csv")
data.head()
data.tail()
```

```python
data.shape
```

```python
data.columns
```

```python
data.isna().sum()
```

```python
data.dtypes
```

```python
data.loc[data["TotalPrice"].isnull(), "TotalPrice"] = data["Price"] * data["Quantity"]
```

```python
data["Date"].fillna(method="ffill",inplace=True)
```

```python
data["Date"].fillna(method="bfill",inplace=True)
```

```python
most_frequent_customer = data["Customer"].mode()[0]
data["Customer"].fillna(most_frequent_customer, inplace=True)   
data.isnull().sum()

```

```python
data["Date"] = pd.to_datetime(data["Date"])    
```

```python
sales_per_category = data.groupby('Category')["TotalPrice"].sum()
print(sales_per_category)
```

```python
sales_per_region = data.groupby('Region')['TotalPrice'].sum()
print(sales_per_region)
```

```python
top_customers = data.groupby('Customer')['TotalPrice'].sum().idxmax()
print(top_customers)
```

```python
expensive_item = data[data['Price'] > 500]
expensive_item
```

```python
electronic_sales = data[data['Category'] == 'Electronics']
electronic_sales
```
